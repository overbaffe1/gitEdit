# Паспорт кристалла
Измеряем [эту модель](crystal.stl). В сохранённом выводе виден результат, даже если Python не запущен. Для пересчёта скачай оба файла в одну папку и открой блокнот в Jupyter.

```python
from pathlib import Path

vertices = [tuple(map(float, line.split()[1:]))
            for line in Path("crystal.stl").read_text().splitlines()
            if line.lstrip().startswith("vertex ")]
faces = [vertices[i:i + 3] for i in range(0, len(vertices), 3)]
edges = {tuple(sorted((a, b))) for face in faces
         for a, b in zip(face, face[1:] + face[:1])}
V, E, F = len(set(vertices)), len(edges), len(faces)
print(f"Вершины: {V}   Рёбра: {E}   Грани: {F}")
print(f"V − E + F = {V - E + F}")
```
